# Exploratory Data Analysis (EDA) - Telco Customer Churn 
**Objective:** Inspect the initial data structure, identify anomalies (data typ errors, missing values), and evaluate the distribution of the target variable ('Churn') to plan our Feature Engineering pipline.

```python
import sys
import os
import pandas as pd

# Setup the path to recognize the src/ directory
sys.path.append(os.path.abspath('..'))
from src.data.loader import DataLoader

csv_file = "../data/raw/telco_churn.csv"
db_file = "../data/processed/telco.db"

# Fetch data from SQLite database
loader = DataLoader(csv_path=csv_file, db_path= db_file)
df= loader.get_data_via_sql("SELECT * FROM customers")

df.head(3)


```

## 1. Data Shape and Data types overview
Check the number of records, features, and the expected vs. actual data types.

```python
print(f"Dataset shape: {df.shape}")
print ("-"*30)
df.info()
```

**Insight:**
- The dataset contains 7043 rows and 21 columns
- Anomaly detected: The 'TotalCharges' column currently has the 'string' data type (needs to be cast to the float in the next step).

```python
# Convert the TotalCharges column to a floating-point numeric type
# The errors= 'coerce' parameter converts all values that cannot be converted to numbers (such as whitespace) into NaN
df['TotalCharges']= pd.to_numeric(df['TotalCharges'],errors='coerce')
```

## 2. Missing Values Check

```python
print("Number of missing values after type conversion: ")
df.isnull().sum()
```

**Insight:**
- There are 11 missing values in TotalCharges 

## 3. Target Distribution (Class Imbalance Check)
Evaluate the proportion of customers who left (yes) vs. stayed (No)

```python
churn_ratio = df['Churn'].value_counts(normalize=True)*100
print(churn_ratio)
```

**Insight:**
- The Churn ratio is approximately 26.5% (Yes) to 73.5% (N0)
- Conclusion: dataset is moderately imbalanced
